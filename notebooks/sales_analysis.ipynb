## Q1) Which region generated the highest total sales?

```python
plt.figure(figsize=(8,5))

sns.barplot(data=region_df,
            x="Region",y="Total_sales",
           hue="Region",legend=False)

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

for container in plt.gca().containers:
    plt.bar_label(container,fmt="%.0f") 
    
plt.savefig("../charts/regional_sales.png")  
plt.show()
```

### Business Insight :
The West region generated the highest total sales at 713,471, while the South region generated the lowest total sales at 388,984. This indicates that the West was the strongest-performing region by sales revenue.

## Q3) Which product category is most profitable in terms of total profit & profit margin?

```python
category_df = pd.read_csv(r"C:\Users\sidhp\Documents\Sales_Performance_Analysis\outputs\03_category_profitability.csv")
category_df.head()
```

```python
category_df.shape
```

```python
category_df.isnull().sum()
```

```python
category_df.duplicated().sum()
```

```python

plt.figure(figsize=(8,6))

sns.barplot(data=category_df,
            x="Category",y="total_profit",hue="Category",palette="viridis",legend=False)

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")

for container in plt.gca().containers:
    plt.bar_label(container,fmt="%.0f")

plt.tight_layout()
plt.savefig("../charts/total_profit_by_category.png")
plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.barplot(data=category_df,
            x="Category",y="profit_margin",hue="Category",palette="viridis",legend=False)

plt.title("Profit Margin by Category")
plt.xlabel("Category")
plt.ylabel("Profit Margin (%)")

for container in plt.gca().containers:
    plt.bar_label(container,fmt="%.2f")

plt.tight_layout()
plt.savefig("../charts/profit_margin_by_category.png")
plt.show()
```

## Business Insight:
Technology was the strongest-performing category, generating the highest total profit of 145,387 and profit margin of 17.39%. Furniture was the weakest-performing category, with the lowest profit margin at 2.32%.

## Q4) How did monthly sales change over time?

```python
monthly_df=pd.read_csv(r"C:\Users\sidhp\Documents\Sales_Performance_Analysis\outputs\04_monthly_sales.csv")
monthly_df.head()
```

```python
monthly_df["Date"]=monthly_df["sales_year"].astype(str)+ "-" + monthly_df["sales_month"].astype(str)
monthly_df.head()
```

```python
monthly_df.info()
```

```python
monthly_df["Date"]=pd.to_datetime(monthly_df["Date"])
monthly_df.head()
```

```python
monthly_df.info()
```

```python
monthly_df.shape
```

```python
monthly_df.isnull().sum()
```

```python
monthly_df.duplicated().sum()
```

```python
plt.figure(figsize=(12,5))

sns.lineplot(data=monthly_df,
             x="Date",y="total_sales",marker="o",linewidth=2)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../charts/monthly_sales_trend.png")
plt.show()
```

## Business Insight:
Monthly sales fluctuated throughout the 2014–2017 period. February generally recorded relatively low sales, while the final quarter generally showed stronger sales performance. The highest monthly sales occurred in November 2017 at 117,383.38, while the lowest occurred in February 2014 at 4,119.82.

## Q5) How does discount affect profit?

```python
discount_df = pd.read_csv(r"C:\Users\sidhp\Documents\Sales_Performance_Analysis\outputs\05_discount_profit.csv")
discount_df.head()
```

```python
discount_df.shape
```

```python
discount_df.isnull().sum()
```

```python
plt.figure(figsize=(8,5))

sns.lineplot(data=discount_df,
             x="Discount",y="avg_profit",legend=False,marker="o",linewidth=2)

plt.title("Average Profit by Discount")
plt.xlabel("Discount")
plt.ylabel("Average Profit")
plt.grid(True)
plt.savefig("../charts/discount_profit.png")
plt.show()
```

## Business Insight:
Average profit was positive at discount levels up to 20% but became negative at discount levels of 30% and above. The lowest average profit occurred at a 50% discount, at -310.70.
